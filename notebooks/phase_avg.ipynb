```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
## If we want to take the mean of the two phases

# make two complex values
img1 = 2+3j
img2 = 2+5j
print(np.angle(img1)*180/np.pi)
print(np.angle(img2)*180/np.pi)

# avg the complex, then get the phase 
print(0.5 * np.angle(img1+img2)*180/np.pi)

# get the two phases, then take the avg
print(0.5 * (np.angle(img1) + np.angle(img2))*180/np.pi)

# are the two above methods give the sae result?
print('For sure they are different, the later one is the correct one')
```
